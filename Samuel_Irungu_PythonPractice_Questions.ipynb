```python
#Q10 Ask the user to input a temperature in Celsius. If the temperature is > 30, print “Hot”, if it’s between 15 and 30 (inclusive), print “Warm”, else print “Cold”.

temperature = float(input("Enter Temperature in Celcius: "))
if(temperature > 30):
  print("Hot")
elif(temperature >= 15):
  print("Warm")
else:
  print("Cold")
```
